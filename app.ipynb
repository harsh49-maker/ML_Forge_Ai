```python
model = LiteLLMModel(
    model_id = "gemini/gemini-3.5-flash",
    api_key = os.getenv("GOOGLE_API_KEY")
)
```

```python
ml_code_agent = CodeAgent(
    tools = [DuckDuckGoSearchTool()],
    additional_authorized_imports=['pandas','numpy','sklearn','json'],
    model=model
)
```

```python
def generate_ml_code(prompt):
    full_prompt = """
    
    You are an AI assistant helping with machine learning code generation.
    complete below 7 steps in a structured manner.
    
    1.**Load the Diabetic dataset** using sklearn.
    2.**split the data**(80% training/ 20% test).
    3.**Train a linear regression model.**
    4.**Find the best 3-feature combination**(lowest RMSE) for the model.
    5.**Find the R2 & adjusted R2** for the model.
    6.**Apply feature engineering** is needed.
    7.**List the top 5 models based on RMSE**.
    8.**Output the results in structured format**.
    9.**Plot the results**.
    
    
    """
    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error :{str(e)}"
    
```

```python
iface = gr.Interface(
    fn = generate_ml_code,
    inputs = "text",
    outputs="text",
    title="ML Code Generator using(Gemini 3.5 Flash)",
    description = "Enter a Machine Learning coding task, and AI will generate a python code for you."
    
)
```

```python
iface.launch()
```
